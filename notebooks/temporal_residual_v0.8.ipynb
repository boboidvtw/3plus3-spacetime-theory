# Temporal Residual v0.8 — Phenomenological Falsifiability Lab
This notebook intentionally adds a hypothetical temporal-identity correction. It is **not derived**, **not evidence**, and **not part of standard QM/GR**. Its purpose is to define observable residuals and consistency tests.

```python
import numpy as np, matplotlib.pyplot as plt
ts=np.linspace(0,10,4001)
def sigmoid(x): return 1/(1+np.exp(-x))
# objective toy identity profiles: A is affected strongly by a merger window; B less strongly
window=sigmoid(7*(ts-3))*sigmoid(7*(7-ts))
I_A=0.85-0.55*window
I_B=0.72-0.22*window
dIA=np.gradient(I_A,ts); dIB=np.gradient(I_B,ts)
# bounded phenomenological functional: state + identity-transition sensitivity
def F(I,dI,a=.6,b=.35): return a*(I-.5)+b*np.tanh(2*dI)
FA=F(I_A,dIA); FB=F(I_B,dIB)
NA_std=np.ones_like(ts); NB_std=np.ones_like(ts)
```

## Local-clock ansatz and relational residual
d tau_i/dt = N_i_std [1 + epsilon F_i]. The construction preserves integrability/transitivity because every clock is defined against the same bookkeeping parameter before ratios are taken.

```python
def rates(eps): return NA_std*(1+eps*FA), NB_std*(1+eps*FB)
def residual(eps):
    na,nb=rates(eps); R=na/nb; Rstd=NA_std/NB_std; return R-Rstd
for eps in [0,1e-4,1e-3,1e-2]: print(eps,np.max(np.abs(residual(eps))))
plt.figure(figsize=(9,4.5))
for eps in [1e-4,1e-3,1e-2]: plt.plot(ts,residual(eps),label=f'eps={eps:g}')
plt.xlabel('t'); plt.ylabel('Delta R_AB'); plt.legend(); plt.grid(alpha=.25); plt.show()
```

## Standard-limit and positivity checks

```python
assert np.allclose(residual(0),0)
for eps in [1e-4,1e-3,1e-2,.1]:
    na,nb=rates(eps); assert np.all(na>0) and np.all(nb>0)
print('standard limit and tested positivity: PASS')
```

## Transitivity test with a third clock

```python
I_C=np.full_like(ts,.62); dIC=np.zeros_like(ts); FC=F(I_C,dIC)
eps=.01; NA=1+eps*FA; NB=1+eps*FB; NC=1+eps*FC
RAB=NA/NB; RBC=NB/NC; RAC=NA/NC
err=np.max(np.abs(RAB*RBC-RAC)); print('max transitivity error=',err)
```

## Relabeling test
Swapping physical labels A and B must invert the relational ratio rather than change the predicted physics.

```python
eps=.01; NA,NB=rates(eps); RAB=NA/NB; RBA=NB/NA
print('max |RAB*RBA-1|=',np.max(np.abs(RAB*RBA-1)))
```

## Small-epsilon scaling
For small epsilon, Delta R_AB is approximately epsilon(F_A-F_B) when standard lapse factors are equal.

```python
shape=FA-FB
for eps in [1e-6,1e-5,1e-4]:
    approx=eps*shape; exact=residual(eps)
    print(eps,np.max(np.abs(exact-approx)))
```

## Null-identical-profile test
If two systems have identical objective identity profiles and standard clock rates, this ansatz must not create a relative anomaly.

```python
for eps in [1e-3,.01,.1]:
    n1=1+eps*FA; n2=1+eps*FA
    assert np.allclose(n1/n2,1)
print('identical-profile relative anomaly: NONE')
```

## Interpretation
The predicted residual is localized around differences in systemhood and identity-transition dynamics. This is only a signal template. A real theory must derive F_i, specify covariance/GR embedding, include environmental clock shifts, and identify an experiment whose standard-model uncertainty is smaller than the predicted residual.
