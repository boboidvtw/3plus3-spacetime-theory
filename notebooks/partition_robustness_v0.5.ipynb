# Systemhood v0.5 — Partition Robustness Lab
Four-qubit interaction graph. The purpose is to test whether dynamical clusters can be recovered from coupling structure rather than observer labels. This is not evidence for emergent physical time.

```python
import numpy as np, itertools, matplotlib.pyplot as plt
J=np.array([[0,1,.05,.02],[1,0,.03,.04],[.05,.03,0,.9],[.02,.04,.9,0.]],float)
K0=.1
def partition_metrics(S):
    S=set(S); E=set(range(len(J)))-S
    win=sum(abs(J[i,j]) for i in S for j in S if i<j)
    wout=sum(abs(J[i,j]) for i in S for j in E)
    A=win/(win+wout+K0)
    Q=(win-wout)/(win+wout+K0)
    return win,wout,A,Q
parts=[]
for r in range(1,4):
    for S in itertools.combinations(range(4),r):
        if 0 not in S: continue # remove complement duplicates
        parts.append((S,*partition_metrics(S)))
for row in sorted(parts,key=lambda x:x[3],reverse=True): print(row)
```

## Visualize autonomy by canonical partition

```python
labels=[str(x[0]) for x in parts]; vals=[x[3] for x in parts]
plt.figure(figsize=(9,4.5)); plt.bar(labels,vals); plt.ylabel('A_partition'); plt.xlabel('candidate S'); plt.xticks(rotation=35); plt.tight_layout(); plt.show()
```

## Robustness under bridge-strength scan
Scale every cross-cluster edge by alpha. A physically meaningful cluster should lose boundary contrast smoothly as cross-coupling becomes comparable to internal coupling.

```python
base=J.copy(); alphas=np.linspace(0,20,201); vals=[]
for a in alphas:
    J=base.copy()
    for i in [0,1]:
        for j in [2,3]: J[i,j]=J[j,i]=a*base[i,j]
    vals.append(partition_metrics((0,1))[2])
J=base
plt.figure(figsize=(8,4.5)); plt.plot(alphas,vals); plt.xlabel('cross-cluster scale alpha'); plt.ylabel('A({0,1})'); plt.grid(alpha=.25); plt.show()
```

## Boundary perturbation test
Compare the intended cluster with nearby partitions obtained by swapping or adding one node.

```python
candidates=[(0,1),(0,2),(0,3),(0,1,2),(0,1,3)]
for S in candidates: print(S,partition_metrics(S))
```

## Interpretation
For this deliberately nonuniform graph, the strong pair {0,1} receives a much higher internal-vs-boundary autonomy score than cross-pairs. This is a graph-structural result, not a temporal result. The next requirement is to add state-dependent quantum mutual information and predictive persistence, then check whether the same boundary remains preferred across dynamics.
